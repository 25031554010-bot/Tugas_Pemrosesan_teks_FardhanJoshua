Web Scraping API

```python
import requests
import pandas as pd
```

```python
api_url = "https://dummyjson.com/products?limit=5"

response = requests.get(api_url)

# Cek apakah request berhasil (Status 200 artinya OK)
if response.status_code == 200:
    print(" Koneksi sukses! Data siap diekstrak.")
else:
    print(f"Gagal mengambil data. Kode status: {response.status_code}")
```

```python
# Mengubah response ke bentuk JSON
data_json = response.json()

# Mengambil list produk dari dalam JSON (cek struktur JSON API-mu untuk nama kunci yang tepat)
daftar_produk = data_json['products']

print(f"Berhasil mengambil {len(daftar_produk)} produk.")
print("Contoh produk pertama:", daftar_produk[0]['title'])
```

```python
# Masukkan data ke dalam DataFrame Pandas
df = pd.DataFrame(daftar_produk)

# Filter kolom-kolom yang penting saja agar rapi
df_rapi = df[['id', 'title', 'price', 'category', 'rating']]

# Menampilkan tabel di Jupyter Notebook
display(df_rapi)

# (Opsional) Simpan data ke dalam file CSV
df_rapi.to_csv("hasil_scrape_api.csv", index=False)
print("Data berhasil disimpan ke 'hasil_scrape_api.csv'")
```
