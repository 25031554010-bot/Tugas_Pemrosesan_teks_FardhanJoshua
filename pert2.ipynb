```python
import re #import regex
text="Pembayaran untuk pesanan #INV-2026-08912 telah berhasil diverifikasi pada pukul 14.30 WIB."
result=re.findall(r"\W\w{3}-\d{4}-\d{5}",text)
print(result)
```

```python
#Ekstraksi Nominal Harga (Rupiah)
text="Total belanjaan Anda adalah Rp 450.000 dengan diskon promo sebesar Rp 50.000."
result=re.findall(r"\w{2} \d+.\d{3}",text)
print(result)
```

```python
# Masking Nomor Kartu Kredit
text="Transaksi menggunakan kartu kredit 4532-1234-5678-9012 telah disetujui."
result=re.findall(r"\d{4}-\d+-\d+-\d+",text)
print(result)
```

```python
# Validasi Kode Promo/Voucher
text="Gunakan kode promo HEMAT50K atau DISKON100 untuk mendapatkan potongan harga."
result=re.findall(r"\w+\d+\w",text)
print(result)
```

```python
#  Identifikasi Metode Pembayaran (E-Wallet/Bank)
text="Metode pembayaran yang dipilih: Transfer Bank BCA (VA: 88012345678)."
result=re.findall(r"Transfer \w+ \w+ \(\w+: \d+\)",text) #tanda () harus diberi tanda \
print(result)
```

```python
#  Ekstraksi Kode SKU Produk
text="Produk Sepatu Lari Kategori A [SKU: SHO-RUN-2026-BLK] tersisa 3 unit."
result=re.findall(r"\w+-\w+-\d+-\w",text)
print(result)
```

```python
# Deteksi Persentase Diskon
text= "Dapatkan diskon hingga 70% untuk produk kecantikan dan diskon ekstra 5% bagi member."
result=re.findall(r"\d+\W",text)
print(result)
```

```python
#  Ekstraksi Dimensi Produk
text="Ukuran paket: 30x20x15 cm dengan berat total 1.5 kg."
result=re.findall(r"\d+\w\d+\w\d+",text)
print(result)
```

```python
# Deteksi Berat Produk
text="Berat bersih item ini adalah 500gr, sedangkan berat packing sekitar 0.75kg."
result=re.findall(r"\d+\W{0,1}\d+\w+",text)
print(result)
```

```python
# Parsing Varian Warna dan Ukuran
text="Kaos Polos Cotton 30s - Warna: Merah Cabai - Size: XL."
result=re.findall(r"\w+: \D+ \D+ - \w+: \w+",text)
print(result)
```

```python
# Ekstraksi Nomor Resi Pengiriman
text= "Paket Anda telah dikirim via JNE dengan no resi JNE102938475661."
result=re.findall(r"\w+\d+",text)
print(result)
```

```python
# Deteksi Kode Pos Indonesia
teks="Alamat tujuan pengiriman: Jl. Mawar No. 12, Jakarta Selatan, 12410."
result=re.findall(r"\d{5}",teks)
print(result)
```

```python
# Deteksi URL/Link Luar yang Dilarang di Chat
text="Silakan transaksi langsung di website kami [www.tokoku-murah.com](https://www.google.com/search?q=https%3A%2F%2Fwww.tokoku-murah.com) ya kak."
result=re.findall(r"\w+\W{0,1}\W*\W*\w+\W+\w+.com[^\s\)]+",text)
print(result)
```

```python
# Ekstraksi Mention ID Pelanggan
text="Customer Support membalas tiket pertanyaan dari customer ID: CUST-9921."
result=re.findall(r"\w+\W\d+",text)
print(result)
```

```python
# Ekstraksi Nilai Rating
text="Produk ini mendapatkan ulasan 4.8 dari 5 bintang berdasarkan 250 pembeli."
result=re.findall(r"\d+\W\d",text)
print(result)
```

```python
import re #import regex
# Deteksi Hashtag Promosi
text="Beli baju baru di promo #FlashSale99 dan #BajuMurah sekarang!"
result=re.findall(r"#\w+\W*",text)
print(result)
```

```python
# Menghapus Tag HTML dari Deskripsi Produk
text="<p>Bahan <b>Katun Premium</b> adem dan nyaman.</p>"
result=re.findall(r"\w{2,} ?\w* ?\w*",text)
print(result)
```

```python
# Ekstraksi Tanggal Ulasan
text="Ulasan ditulis oleh Ahmad pada tanggal 12/08/2026."
result=re.findall(r"\d+/\d+/\d+",text)
print(result)
```

```python
import re #import regex
# Deteksi Sentimen Kata Kunci Positif
text="Pengiriman sangat cepat, kualitas ori dan seller ramah!"
result=re.findall(r"cepat|ori|ramah",text)
print(result)
```

```python
# Ekstraksi Alamat IP Pengguna
text="Aktivitas login mencurigakan terdeteksi dari IP 192.168.1.105 pada lokasi Jakarta."
result=re.findall(r"IP \d+.\d+.\d+.\d+",text)
print(result)
```

```python
# Parsing Log Waktu Server
text="[2026-09-02 08:30:15] ERROR: Gagal memproses transaksi ID 992."
result=re.findall(r"\d+-\d+-\d+ \d+:\d+:\d+",text)
print(result)
```

```python
# Ekstraksi HTTP Status Code
text="API Payment Gateway merespons dengan status POST /checkout HTTP/1.1 200 OK."
result=re.findall(r"\w+ /\w+ \w+/\d\W?\d \d+ \w+",text)
print(result)
```

```python
# Deteksi Percobaan SQL Injection di Kolom Pencarian
text="User mencari keyword: baju pria' OR '1'='1"
result=re.findall(r"\w+ \w+'|\d'",text)
print(result)
```

```python
# Parsing Slug URL Produk
text="Halaman diakses: https://tokopedia.com/p/sepatu-sneakers-pria-white-edition"
result=re.findall(r"\w+\W+\w+\W\w+/\w/\w+\W?\w+\W?\w+\W?\w+\W?\w+",text)
print(result)
```

```python
# deteksi Batas Waktu Bayar (Countdown/Timer)
text="Selesaikan pembayaran Anda dalam 01:45:30 sebelum pesanan dibatalkan."
result=re.findall(r"\d+:\d+:\d+",text)
print(result)
```

```python
# Parsing Jumlah Stok (Inventory Quantity)
text="Stok Gudang A: 150 pcs | Stok Gudang B: 20 pcs."
result=re.findall(r"\w: \d+",text)
print(result)
```

```python
# Ekstraksi Kode Rak/Lokasi Bin Gudang
text="Ambil barang di Lokasi Rak: RAK-A3-BAY12."
result=re.findall(r"\w+: \w+-\w+-\w+",text)
print(result)
```

```python
# Deteksi Status Pesanan
text="Pesanan #1092 berubah status dari [MENUNGGU_PAYMENT] menjadi [DISELESAIKAN]."
result=re.findall(r"\[\w+_?\w+\]",text)
print(result)
```

```python
# Parsing Berat Total & Biaya Ongkir
text="Berat Total: 3kg - Biaya Pengiriman: Rp 27.000."
result=re.findall(r"\d\wg|\w+ \d+\W\d+",text)
print(result)
```

```python
#  Membersihkan Multiple White Space pada Deskripsi
text="Baju    kaos   pria     lengan panjang."
result=re.findall(r"\w+\S+\w+",text)
print(result)
```
