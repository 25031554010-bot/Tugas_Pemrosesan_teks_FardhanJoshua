```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

# Menyiapkan header agar server tidak mengabaikan parameter ?page=
headers = {
    'User-Agent': (
        'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML,'
        ' like Gecko) Chrome/120.0.0.0 Safari/537.36'
    )
}

data_produk = []

# Loop untuk mengambil data dari halaman 1 sampai 3
for page in range(1, 4):
  url = f'https://sandbox.oxylabs.io/products?page={page}'
  print(f'Sedang mengambil data dari halaman {page}...')

  # Kirim request dengan parameter headers
  response = requests.get(url, headers=headers)

  if response.status_code == 200:
    soup = BeautifulSoup(response.text, 'html.parser')

    # Oxylabs menggunakan class "product-card"
    products = soup.find_all('div', class_='product-card')

    for item in products:
      # 1. Judul (di situs ini menggunakan tag h4)
      title_elem = item.find('h4')
      title = title_elem.get_text(strip=True) if title_elem else 'N/A'

      deskrip_elem = item.find(['p'], class_='description css-cput12 eag3qlw5')
      deskrip = deskrip_elem.get_text(strip=True) if deskrip_elem else 'N/A'

      # 2. Genre / Kategori
      genre_elem = item.find_all(['span', 'p'], class_='css-1pewyd6')
      lol = [g.get_text(strip=True) for g in genre_elem]
      genre_text = ', '.join(lol) if lol else 'N/A'

      # 3. Harga (di situs ini menggunakan class "price-wrapper")
      price_elem = item.find('div', class_='price-wrapper')
      price = price_elem.get_text(strip=True) if price_elem else 'N/A'

      # Memasukkan data ke dalam list utama
      data_produk.append({
          'Halaman': page,
          'Judul': title,
          'Genre': genre_text,
          'deskripsi' : deskrip,
        #   'Harga': price,
      })
  else:
    print(
        f'Gagal mengakses halaman {page}. Status code: {response.status_code}'
    )

# Mengubah data dari ketiga halaman menjadi DataFrame
df = pd.DataFrame(data_produk)

# Menampilkan jumlah total baris dan isi tabelnya
print(f'\nTotal data berhasil diambil: {len(df)} item')
print(df)
```

```python
import pandas as pd
import requests
from bs4 import BeautifulSoup

headers = {
    'User-Agent': (
        'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML,'
        ' like Gecko) Chrome/120.0.0.0 Safari/537.36'
    )
}

data_produk = []

for page in range(1, 4):
    url = f'https://sandbox.oxylabs.io/products?page={page}'
    print(f'Sedang mengambil data dari halaman {page}...')

    response = requests.get(url, headers=headers)

    if response.status_code == 200:
        soup = BeautifulSoup(response.text, 'html.parser')
        products = soup.find_all('div', class_='product-card')

        for item in products:
            title_elem = item.find('h4')
            title = title_elem.get_text(strip=True) if title_elem else 'N/A'

            deskrip_elem = item.find(['p'], class_='description css-cput12 eag3qlw5')
            deskrip = deskrip_elem.get_text(strip=True) if deskrip_elem else 'N/A'

            genre_elem = item.find_all(['span', 'p'], class_='css-1pewyd6')
            lol = [g.get_text(strip=True) for g in genre_elem]
            genre_text = ', '.join(lol) if lol else 'N/A'

            price_elem = item.find('div', class_='price-wrapper')
            price = price_elem.get_text(strip=True) if price_elem else 'N/A'

            data_produk.append({
                'Halaman': page,
                'Judul': title,
                'Genre': genre_text,
                'Deskripsi': deskrip,
                'Harga': price
            })
    else:
        print(f'Gagal mengakses halaman {page}. Status code: {response.status_code}')

# 1. Mengubah data menjadi DataFrame
df = pd.DataFrame(data_produk)

# 2. Menyimpan DataFrame ke file CSV
nama_file = 'produk_oxylabs.csv'
df.to_csv(nama_file, index=False, encoding='utf-8-sig')

print(f'\nData berhasil disimpan ke file: {nama_file}')
```
